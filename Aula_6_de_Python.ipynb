# Desafio Sprint 6 - Sistema Orientado a Objetos

Sistema desenvolvido em Python utilizando Programação Orientada a Objetos (POO).

O projeto demonstra:

- Classes e objetos
- Construtores
- Encapsulamento
- @property
- Herança
- super()
- Polimorfismo
- Classe abstrata
- Métodos especiais (__str__, __repr__ e __eq__)
- Tratamento de exceções
- Composição

```python
from abc import ABC, abstractmethod
```

```python
class Produto(ABC):

    def __init__(self, nome, preco):
        self.nome = nome
        self.preco = preco

    @property
    def nome(self):
        return self._nome

    @nome.setter
    def nome(self, valor):

        if not isinstance(valor, str) or valor.strip() == "":
            raise ValueError("O nome não pode ser vazio.")

        self._nome = valor

    @property
    def preco(self):
        return self._preco

    @preco.setter
    def preco(self, valor):

        if valor < 0:
            raise ValueError("O preço não pode ser negativo.")

        self._preco = float(valor)

    @abstractmethod
    def calcular_preco_final(self):
        pass

    def __str__(self):
        return f"{self.nome} - R$ {self.preco:.2f}"

    def __repr__(self):
        return f"Produto(nome='{self.nome}', preco={self.preco})"

    def __eq__(self, outro):

        if not isinstance(outro, Produto):
            return False

        return self.nome == outro.nome and self.preco == outro.preco
```

```python
class ProdutoFisico(Produto):

    def __init__(self, nome, preco, frete):
        super().__init__(nome, preco)
        self.frete = frete

    @property
    def frete(self):
        return self._frete

    @frete.setter
    def frete(self, valor):

        if valor < 0:
            raise ValueError("O frete não pode ser negativo.")

        self._frete = float(valor)

    def calcular_preco_final(self):

        return self.preco + self.frete

    def __str__(self):

        return (
            f"{self.nome} | "
            f"Preço: R$ {self.preco:.2f} | "
            f"Frete: R$ {self.frete:.2f} | "
            f"Total: R$ {self.calcular_preco_final():.2f}"
        )

    def __repr__(self):

        return (
            f"ProdutoFisico("
            f"nome='{self.nome}', "
            f"preco={self.preco}, "
            f"frete={self.frete})"
        )
```

```python
class ProdutoDigital(Produto):

    def __init__(self, nome, preco, tamanho_mb):
        super().__init__(nome, preco)
        self.tamanho_mb = tamanho_mb

    @property
    def tamanho_mb(self):
        return self._tamanho_mb

    @tamanho_mb.setter
    def tamanho_mb(self, valor):

        if valor <= 0:
            raise ValueError("O tamanho deve ser maior que zero.")

        self._tamanho_mb = float(valor)

    def calcular_preco_final(self):

        return self.preco

    def __str__(self):

        return (
            f"{self.nome} | "
            f"Preço: R$ {self.preco:.2f} | "
            f"Tamanho: {self.tamanho_mb:.1f} MB | "
            f"Total: R$ {self.calcular_preco_final():.2f}"
        )

    def __repr__(self):

        return (
            f"ProdutoDigital("
            f"nome='{self.nome}', "
            f"preco={self.preco}, "
            f"tamanho_mb={self.tamanho_mb})"
        )
```

```python
class Cliente:

    def __init__(self, nome, email):
        self.nome = nome
        self.email = email
        self.produtos = []

    @property
    def nome(self):
        return self._nome

    @nome.setter
    def nome(self, valor):

        if not isinstance(valor, str) or valor.strip() == "":
            raise ValueError("O nome não pode ser vazio.")

        self._nome = valor

    @property
    def email(self):
        return self._email

    @email.setter
    def email(self, valor):

        if "@" not in valor or "." not in valor:
            raise ValueError("E-mail inválido.")

        self._email = valor

    def adicionar_produto(self, produto):

        if not isinstance(produto, Produto):
            raise TypeError("O objeto precisa ser um produto.")

        self.produtos.append(produto)

    def __str__(self):

        return (
            f"Cliente: {self.nome} | "
            f"E-mail: {self.email} | "
            f"Produtos: {len(self.produtos)}"
        )

    def __repr__(self):

        return (
            f"Cliente("
            f"nome='{self.nome}', "
            f"email='{self.email}')"
        )
```

```python
from abc import ABC, abstractmethod

class Produto(ABC):
    def __init__(self, nome: str, preco: float):
        self._nome = nome
        self.preco = preco  # Aciona o setter para validar o preço inicial

    # Encapsulamento do Nome
    @property
    def nome(self) -> str:
        return self._nome

    @nome.setter
    def nome(self, valor: str):
        self._nome = valor

    # Encapsulamento do Preço com Validação
    @property
    def preco(self) -> float:
        return self._preco

    @preco.setter
    def preco(self, valor: float):
        if valor < 0:
            raise ValueError("O preço não pode ser negativo.")
        self._preco = float(valor)

    # Método Abstrato (Polimorfismo)
    @abstractmethod
    def calcular_preco_final(self) -> float:
        pass

    # Métodos Especiais
    def __str__(self) -> str:
        return f"{self.nome} - R$ {self.preco:.2f}"

    def __repr__(self) -> str:
        return f"{self.__class__.__name__}(nome='{self.nome}', preco={self.preco})"

    def __eq__(self, outro) -> bool:
        if not isinstance(outro, Produto):
            return False
        return self.nome == outro.nome and self.preco == outro.preco


class ProdutoFisico(Produto):
    def __init__(self, nome: str, preco: float, frete: float):
        super().__init__(nome, preco)
        self._frete = frete

    @property
    def frete(self) -> float:
        return self._frete

    @frete.setter
    def frete(self, valor: float):
        if valor < 0:
            raise ValueError("O frete não pode ser negativo.")
        self._frete = float(valor)

    def calcular_preco_final(self) -> float:
        return self.preco + self.frete


class ProdutoDigital(Produto):
    def __init__(self, nome: str, preco: float, tamanho_mb: float):
        super().__init__(nome, preco)
        self._tamanho_mb = tamanho_mb

    @property
    def tamanho_mb(self) -> float:
        return self._tamanho_mb

    @tamanho_mb.setter
    def tamanho_mb(self, valor: float):
        if valor <= 0:
            raise ValueError("O tamanho em MB deve ser maior que zero.")
        self._tamanho_mb = float(valor)

    def calcular_preco_final(self) -> float:
        return self.preco


class Cliente:
    def __init__(self, nome: str, email: str):
        self._nome = nome
        self._email = email
        self._produtos = []

    @property
    def nome(self) -> str:
        return self._nome

    @property
    def email(self) -> str:
        return self._email

    @property
    def produtos(self) -> list:
        return self._produtos

    def adicionar_produto(self, produto: Produto):
        if isinstance(produto, Produto):
            self._produtos.append(produto)
        else:
            raise TypeError("Só é possível adicionar objetos que herdam de Produto.")
```
